## preparing the data

```python
import spacy
import pandas as pd
from spacy import displacy

nlp_sm = spacy.load("en_core_web_sm")
nlp_md = spacy.load("en_core_web_md")

def extract_entities(text, model):
    doc = model(text)
    entities = []
    for ent in doc.ents:
        entities.append((ent.text, ent.label_))
    return entities

sample_news = "Apple is looking at buying a startup in the United Kingdom for $1 billion. Tim Cook is visiting London next week."

print("Entities using Small Model:")
print(extract_entities(sample_news, nlp_sm))
```

## displacy

```python
displacy.render(nlp_sm(sample_news), style="ent", jupyter=True)
```

```python
def compare_models(text):
    doc_sm = nlp_sm(text)
    doc_md = nlp_md(text)
    
    print(f"{'Text':<20} | {'Small Model':<15} | {'Medium Model':<15}")
    print("-" * 55)
    
    for ent in doc_md.ents:
        sm_label = next((e.label_ for e in doc_sm.ents if e.text == ent.text), "Not Found")
        print(f"{ent.text:<20} | {sm_label:<15} | {ent.label_:<15}")

compare_models(sample_news)
```
